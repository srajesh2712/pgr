```python
import pandas as pd

# Path to your CSV file
file_path = "data/hly275.csv"

# Met Éireann files have metadata at the top.
# We find the line number where the actual header 'date' appears.
header_row = 0
with open(file_path, "r") as f:
    for i, line in enumerate(f):
        if line.strip().startswith("date"):
            header_row = i
            break

# Load the CSV, skipping the metadata legend lines
df = pd.read_csv(file_path, skiprows=header_row)

# Strip any accidental whitespace in column names
df.columns = df.columns.str.strip()

# Convert the 'date' column into a proper pandas Datetime object
df["date"] = pd.to_datetime(df["date"], format="%d-%b-%Y %H:%M")

# Sort chronologically
df = df.sort_values("date").reset_index(drop=True)

# Display basic info and first few rows
print("Data successfully loaded!")
print(f"Total observations: {len(df)}")
print(f"Date range: {df['date'].min()} to {df['date'].max()}")
print("\nFirst 5 rows:")
print(df[["date", "temp", "rhum", "msl", "wdsp", "rain"]].head())
```
