# calculate PGF

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


def load_met_eireann_csv(filepath):
    # Locate where data rows actually start
    with open(filepath, "r") as f:
        lines = f.readlines()

    header_line = 0
    for idx, line in enumerate(lines):
        if "date" in line.lower() and "rain" in line.lower():
            header_line = idx
            break

    # Read CSV skipping metadata
    df = pd.read_csv(filepath, skiprows=header_line)

    # Clean header strings
    df.columns = df.columns.str.strip().str.lower()

    # Handle duplicate 'ind' column names safely by assigning unique names
    cols = []
    count = {}
    for col in df.columns:
        if col in count:
            count[col] += 1
            cols.append(f"{col}_{count[col]}")
        else:
            count[col] = 0
            cols.append(col)
    df.columns = cols

    # Parse date column safely ('21-nov-2024 00:00' format)
    df["date"] = pd.to_datetime(df["date"], format="%d-%b-%Y %H:%M")

    # Coerce target columns to float
    df["msl"] = pd.to_numeric(df["msl"], errors="coerce")
    df["wdsp"] = pd.to_numeric(df["wdsp"], errors="coerce")

    return df[["date", "msl", "wdsp"]]


# 1. Load Datasets
df_malin = load_met_eireann_csv("data/hly1575_subset.csv")
df_knock = load_met_eireann_csv("data/hly4935_subset.csv")

# 2. Inner Merge on exact matching datetime
merged_df = pd.merge(
    df_malin, df_knock, on="date", suffixes=("_malin", "_knock")
).dropna(subset=["msl_malin", "msl_knock"])

# 3. Calculate Hourly Pressure Gradient (|Malin MSL - Knock MSL|)
merged_df["delta_p"] = (merged_df["msl_malin"] - merged_df["msl_knock"]).abs()

# Filter for Storm Bert Window (Nov 21 - Nov 27, 2024)
storm_df = merged_df[
    (merged_df["date"] >= "2024-11-21") & (merged_df["date"] <= "2024-11-27")
].copy()

# 4. Dual-Axis Plotting
fig, ax1 = plt.subplots(figsize=(14, 6))

# Axis 1: Pressure Difference (Proxy for PGF)
color1 = "crimson"
ax1.set_xlabel("Date & Time (UTC)", fontsize=11, fontweight="bold")
ax1.set_ylabel(
    "Pressure Difference: |Malin Head - Knock| (hPa)",
    color=color1,
    fontsize=11,
    fontweight="bold",
)
line1 = ax1.plot(
    storm_df["date"],
    storm_df["delta_p"],
    color=color1,
    linewidth=2.2,
    label="Pressure Gradient (ΔP)",
)
ax1.tick_params(axis="y", labelcolor=color1)

# Axis 2: Wind Speeds
ax2 = ax1.twinx()
color2 = "navy"
ax2.set_ylabel(
    "Wind Speed (knots)", color=color2, fontsize=11, fontweight="bold"
)
line2 = ax2.plot(
    storm_df["date"],
    storm_df["wdsp_malin"],
    color=color2,
    linewidth=1.8,
    linestyle="--",
    label="Malin Head Wind Speed (kt)",
)
line3 = ax2.plot(
    storm_df["date"],
    storm_df["wdsp_knock"],
    color="teal",
    linewidth=1.5,
    linestyle=":",
    label="Knock Airport Wind Speed (kt)",
)
ax2.tick_params(axis="y", labelcolor=color2)

# Combine Legends
lines = line1 + line2 + line3
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc="upper left")

plt.title(
    "Storm Bert: Hourly Pressure Gradient (Malin Head vs Knock) vs Wind Speed Response",
    fontsize=13,
    fontweight="bold",
    pad=12,
)
ax1.grid(True, linestyle=":", alpha=0.6)
fig.tight_layout()
plt.show()

# 5. Output Correlation Summary
corr_malin = storm_df["delta_p"].corr(storm_df["wdsp_malin"])
corr_knock = storm_df["delta_p"].corr(storm_df["wdsp_knock"])

print("=== STRENGTH OF PRESSURE GRADIENT RELATIONSHIP ===")
print(f"Pearson Correlation (ΔP vs Malin Head Wind): {corr_malin:.3f}")
print(f"Pearson Correlation (ΔP vs Knock Airport Wind): {corr_knock:.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Load processed storm DataFrame (from previous step)
# Ensure storm_df contains: 'date', 'msl_malin', 'msl_knock', 'wdsp_malin', 'wdsp_knock'

# --- Physical Parameters ---
# Distance between Malin Head and Knock Airport (~187 km)
DISTANCE_M = 187000.0

# Mean Latitude (~54.6 degrees North)
LATITUDE_RAD = np.radians(54.6)
OMEGA = 7.2921e-5  # Earth rotation rate (rad/s)
f = 2 * OMEGA * np.sin(LATITUDE_RAD)  # Coriolis parameter (~0.000119 s^-1)

RHO = 1.225  # Standard air density at surface (kg/m^3)

# 1. Compute Pressure Difference in Pascals (hPa * 100)
delta_p_pa = (storm_df["msl_malin"] - storm_df["msl_knock"]).abs() * 100.0

# 2. Geostrophic Wind Calculation (m/s to knots)
storm_df["v_geostrophic_ms"] = delta_p_pa / (RHO * f * DISTANCE_M)
storm_df["v_geostrophic_kt"] = storm_df["v_geostrophic_ms"] * 1.94384

# --- Visualization ---
fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(
    storm_df["date"],
    storm_df["v_geostrophic_kt"],
    color="crimson",
    linewidth=2,
    label="Geostrophic Wind (Theoretical Vg)",
)
ax.plot(
    storm_df["date"],
    storm_df["wdsp_malin"],
    color="navy",
    linewidth=1.8,
    linestyle="--",
    label="Malin Head Observed Wind (kt)",
)
ax.plot(
    storm_df["date"],
    storm_df["wdsp_knock"],
    color="teal",
    linewidth=1.5,
    linestyle=":",
    label="Knock Airport Observed Wind (kt)",
)

ax.set_xlabel("Date & Time (UTC)", fontweight="bold")
ax.set_ylabel("Wind Speed (knots)", fontweight="bold")
ax.set_title(
    "Storm Bert: Geostrophic Wind Speed vs Observed Surface Winds",
    fontsize=13,
    fontweight="bold",
    pad=12,
)
ax.legend(loc="upper left")
ax.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()

# Ratio comparison (Boundary Layer Friction effect)
avg_malin_ratio = (
    storm_df["wdsp_malin"] / storm_df["v_geostrophic_kt"]
).mean() * 100
avg_knock_ratio = (
    storm_df["wdsp_knock"] / storm_df["v_geostrophic_kt"]
).mean() * 100

print(f"Malin Head Surface Wind is ~{avg_malin_ratio:.1f}% of Geostrophic Wind")
print(f"Knock Airport Surface Wind is ~{avg_knock_ratio:.1f}% of Geostrophic Wind")
```
